```python
import numpy as np
```

```python
array_1d = np.array([1, 2, 3, 4, 5])
print(array_1d.ndim) # Número de dimensiones
print(array_1d.shape) # Forma del array (número de elementos en cada dimensión)
print(array_1d)
```

```python
array_2d = np.array([[1, 2, 3], [4, 5, 6]])
print(array_2d.ndim) # Número de dimensiones    
print(array_2d.shape) # Forma del array (número de elementos en cada dimensión)
print(array_2d)
# Matrizes, tabelas de dados, imagens em escala de cinza
```

```python
array_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(array_3d.ndim) # Número de dimensiones    
print(array_3d.shape) # Forma del array (número de elementos en cada dimensión)
print(array_3d)
```
